```python
import requests

url = "https://scrapifydatalabs.com/playground/static-products/"

response = requests.get(url)

print(response.status_code)
```

```python
with open("products.txt", "w", encoding="utf-8") as file:
    file.write(response.text)
```

```python
from bs4 import BeautifulSoup
with open("products.txt","r" , encoding="utf-8") as file:
    html= file.read()
soup = BeautifulSoup(html,"html.parser")
```

```python
products= soup.find_all("article" , class_="product-card")
```

```python
products_data = []

for product in products:
    name = product.find("h3", class_="product-title").text.strip()
    price = product.find("div", class_="product-price").text.strip()
    rating = product.find("div", class_="product-rating").text.strip()
    stock = product.find("div", class_="product-stock").text.strip()

    data = {
        "name": name,
        "price": price,
        "rating": rating,
        "stock": stock
    }

    products_data.append(data)
```

```python
print(products_data)
```

```python
import pandas as pd

df = pd.DataFrame(products_data)

df
```

```python
df.to_csv("products.csv", index=False)

print("Data saved successfully!")
```
